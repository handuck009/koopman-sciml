# OAT15A EDMD-DL Following Qianxiao Li et al. (2017)

The dictionary contains fixed constant and POD-coordinate observables plus trainable outputs from a three-hidden-layer tanh neural network. Training alternates between an analytic Koopman update and a neural-dictionary gradient update.
